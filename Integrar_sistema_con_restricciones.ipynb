```python
from typing import List


def ordenar_por_seleccion(temperaturas: List[float]) -> List[float]:
    """Ordena una lista de temperaturas flotantes de menor a mayor

    utilizando el algoritmo de Selección (Selection Sort).
    """
    lista_ordenada = temperaturas.copy()
    longitud = len(lista_ordenada)

    for i in range(longitud - 1):
        indice_minimo = i
        for j in range(i + 1, longitud):
            if lista_ordenada[j] < lista_ordenada[indice_minimo]:
                indice_minimo = j

        if indice_minimo != i:
            # Intercambio idiomático de Python
            lista_ordenada[i], lista_ordenada[indice_minimo] = (
                lista_ordenada[indice_minimo],
                lista_ordenada[i],
            )

    return lista_ordenada


def main() -> None:
    # Dataset inicial de la estación meteorológica
    temperaturas_estacion = [25.4, 12.8, 35.1, 4.2, 18.5]

    print("Dataset original:", temperaturas_estacion)

    # Procesamiento / Ordenamiento
    resultado_ordenado = ordenar_por_seleccion(temperaturas_estacion)

    print("Dataset ordenado: ", resultado_ordenado)


if __name__ == "__main__":
    main()
```
