```python
from typing import Dict, List, Optional, Union


def depurar_datos(datos_brutos: List[Optional[Union[int, float]]]) -> List[float]:
    """Filtra valores nulos y negativos de la lista, retornando flotantes válidos."""
    return [
        float(v) for v in datos_brutos if v is not None and v >= 0
    ]


def calcular_estadisticas(
    datos: List[float],
) -> Optional[Dict[str, float]]:
    """Calcula la media, varianza y desviación estándar de una lista de datos."""
    n = len(datos)
    if n == 0:
        return None

    media = sum(datos) / n
    varianza = sum((x - media) ** 2 for x in datos) / n
    desviacion_estandar = varianza**0.5

    return {
        "media": media,
        "varianza": varianza,
        "desviacion_estandar": desviacion_estandar,
    }


# --- Bloque Principal de Ejecución ---
if __name__ == "__main__":
    datos_XYZ = [12.5, -999, 15.0, None, 18.2, 20.1, -5, 14.8]

    # Paso 1: Depuración de datos
    datos_depurados = depurar_datos(datos_XYZ)

    # Paso 2: Cálculo estadístico
    resultado = calcular_estadisticas(datos_depurados)

    print("Datos originales:", datos_XYZ)
    print("Datos depurados: ", datos_depurados)
    print("Resultado:       ", resultado)
```
